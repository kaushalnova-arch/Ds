5b. Write a python code to show statistical information on given data set. 

```python
import statistics

print(statistics.mean([1, 3, 5, 7, 9, 11, 13]))
print(statistics.mean([1, 3, 5, 7, 9, 11]))
print(statistics.mean([-11, 5.5, -3.4, 7.1, -9, 22]))
```

statistics.median()

```python
import statistics

print(statistics.median([10, 9, 3, 8, 2, 1, 11]))
print(statistics.median([1, 3, 5, 7, 9, 11]))
print(statistics.median([-11, 5.5, -3.4, 7.1, -9, 22]))
```

statistics.mode()

```python
import statistics

print(statistics.mode([1, 5, 3, 4, 7, 3, 9, 11]))
print(statistics.mode([1, 1, 9, -5, 7, -9, 11]))
print(statistics.mode(['Ajay', 'Sunny', 'Amit', 'rani']))
```

statistics.stdev()

```python
import statistics

print(statistics.stdev([1, 3, 5, 6, 4, 15]))
print(statistics.stdev([2, 2.5, 1.75, 3.1, 1.75, 2.8]))
print(statistics.stdev([-11, 5.5, -3.1, 7.1]))
print(statistics.stdev([1, 30, 50, 200]))
```

statistics.variance()

```python
import statistics

print(statistics.variance([1, 3, 6, 7, 9, 11]))
print(statistics.variance([2, 2.5, 1.44, 3.1, 1.75, 2.8]))
print(statistics.variance([-11, 5.5, -3.4, 7.1]))
print(statistics.variance([1, 30, 50, 500]))
```

5d. Write a python code to demonstrate filter pandas series with Boolean arrays.

```python
import pandas as pd
dict = {'name':["kaushal", "Yaduvanshi", "sudhir", "Geeku"],
        'degree': ["MBA", "BCA", "M.Tech", "MBA"],
        'score':[90, 40, 80, 98]}

df = pd.DataFrame(dict, index = [True, False, True, False])
print(df.loc[True])
```
